### Create Indexes

```python
# QUERY OPTIMISATION - INDEX CREATION

db["orders"].create_index("delivery_status")
db["complaints"].create_index("severity")
db["orders"].create_index("hub_id")
db["orders"].create_index("driver_id")

print("Indexes created successfully")
```

```python
print("=== EXPLAIN PLAN: Failed Deliveries ===")

explain = db["orders"].find(
    {"delivery_status": "Failed"}
).explain()

stats = explain["executionStats"]

print("Documents Examined:", stats["totalDocsExamined"])
print("Documents Returned:", stats["nReturned"])
print("Execution Time (ms):", stats["executionTimeMillis"])

print("Winning Plan:",
      explain["queryPlanner"]["winningPlan"]["stage"])
```

```python
# LIST INDEXES

print("=== Orders Indexes ===")
for i in db["orders"].list_indexes():
    print(i)

print("\n=== Complaints Indexes ===")
for i in db["complaints"].list_indexes():
    print(i)
```

```python
# PERFORMANCE TESTING

import time

# delivery_status
start = time.time()
list(db["orders"].find({"delivery_status": "Failed"}))
end = time.time()
print("delivery_status query time:", round((end-start)*1000, 2), "ms")

# severity
start = time.time()
list(db["complaints"].find({"severity": "High"}))
end = time.time()
print("severity query time:", round((end-start)*1000, 2), "ms")

# hub_id
start = time.time()
list(db["orders"].find({"hub_id": "H01"}))
end = time.time()
print("hub_id query time:", round((end-start)*1000, 2), "ms")
```

## Query Optimisation Justification

Indexes were created on frequently queried fields:

- delivery_status (orders)
  Used in analysing failed deliveries
  Reduces full collection scan

- severity (complaints)
  Used to identify high priority issues
  Enables faster filtering

- hub_id (orders)
  Supports hub-level performance analysis

- driver_id (orders)
  Supports driver tracking and evaluation

Without indexes:
MongoDB scans all documents (collection scan)

With indexes:
MongoDB directly accesses matching records

Even though dataset is small, indexing improves
scalability and prepares system for real-world usage.
